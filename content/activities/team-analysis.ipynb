# Lesson: Team Analysis Coding Challenge

In this activity, we will use Python to compare the performance of multiple NBA teams. First we will import the data from the 2025-2026 season.

```python
import piplite, pyodide
await piplite.install(['pandas', 'plotly', 'nbformat', 'statsmodels'])
import pandas as pd
import plotly.express as px
df = pd.read_csv(pyodide.http.open_url('https://raw.githubusercontent.com/Data-Dunkers/data/refs/heads/main/NBA/team/nba_team_stats_2025-2026.csv'))
df
```

## Visualizing with Plotly

We will use `plotly.express` to create a grouped bar chart.

```python
metrics = ['PTS', 'REB', 'AST', 'STL', 'BLK']  # the metrics we want to compare
px.bar(df, x='Team', y=metrics, barmode='group', title='NBA Team Comparison (2025-2026)')
```

If we'd like to include just a selection of teams, we can first list all of the teams.

```python
df['teams']
```

Then we can filter the dataframe for just the teams we would like to show.

```python
teams = ['Indiana Pacers', 'Toronto Raptors', 'Boston Celtics']
filtered_df = df[df['Team'].isin(teams)]
metrics = ['PTS', 'REB', 'AST', 'STL', 'BLK']
px.bar(filtered_df, x='Team', y=metrics, barmode='group', title='Selected NBA Team Comparison (2025-2026)')
```
