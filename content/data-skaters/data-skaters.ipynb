# Data Skaters - Data Science

This Jupyter notebook is an introduction to data science in Python using our data and NHL data.

## Importing Our Data

```python
import piplite, pyodide
await piplite.install(['pandas', 'plotly', 'nbformat', 'statsmodels'])
import requests
import pandas as pd
import plotly.express as px

traits_url = 'https://api.datadunkers.ca/api/collections/data_skaters_traits/records'
response = requests.get(traits_url, params={'perPage':500, 'page':1})
all_traits = pd.DataFrame(response.json()['items'])
display(all_traits)

shots_url = 'https://api.datadunkers.ca/api/collections/data_skaters_shots/records'
first_page = requests.get(shots_url, params={'perPage':500, 'page':1}).json()
item_list = first_page['items']
for n in range(2, first_page['totalPages'] + 1):
  item_list.extend(requests.get(shots_url, params={'perPage':500, 'page':n}).json()['items'])
all_shots = pd.DataFrame(item_list)
display(all_shots)
```

### Filter for Today

```python
today = '2026-04-12'

traits = all_traits[all_traits['created'].str.contains(today)]
shots = all_shots[all_shots['created'].str.contains(today)]
display(traits)
```

### Visualize

```python
px.scatter(traits, x='resting_heart_rate', y='reaction_time_ms', trendline='ols', title='Reaction Time vs Resting Heart Rate')
```

```python
group_success_rate = shots.groupby('group')['shots_made'].mean().reset_index()
px.bar(group_success_rate, x='group', y='shots_made', title='Shots Made by Group')
```

```python
px.imshow(traits.corr(numeric_only=True), title='Correlation Matrix', height=800, text_auto='.2f')
```

## NHL Data

Next let's look at data from the Winnipeg Jets 2025-2026 season.

```python
url = 'https://raw.githubusercontent.com/Data-Dunkers/data/refs/heads/main/NHL/team_players/2025-2026/WPG_2025-2026_players.csv'
df_jets = pd.read_csv(pyodide.http.open_url(url))
df_jets
```

Now we can find the top ten players, by points, on the team.

```python
df_sorted = df_jets.sort_values(by='PTS', ascending=False)
df_sorted.head(10)
```

Then we can use Plotly Express to create a bar chart of the top ten players and their points.

```python
px.bar(df_sorted.head(10), x='Name', y='PTS',  title='Winnipeg Jets Top 10 Point Leaders (2025-2026)', labels={'Name': 'Player Name', 'PTS': 'Points'})
```

Next let's look at a larger dataset of all player statistics from the 2025-2026 season.

```python
url = 'https://raw.githubusercontent.com/Data-Dunkers/data/refs/heads/main/NHL/player/nhl_player_stats_2025-2026.csv'
df_nhl = pd.read_csv(pyodide.http.open_url(url))
df_nhl
```

Then we can create a scatterplot of goals versus assists.

```python
px.scatter(df_nhl, x='G', y='A', hover_name='Name', trendline='ols', title='NHL Goals vs Assists (2025-2026)', labels={'G': 'Goals', 'A': 'Assists'})
```

We can also filter the data set to just show the Winnipeg Jets.

```python
df_jets = df_nhl[df_nhl['Team'] == 'WPG']
df_jets
```

And then create a scatteplot of just the Jets data.

```python
px.scatter(df_jets, x='G', y='A', hover_name='Name', trendline='ols', title='Jets Goals vs Assists (2025-2026)', labels={'G': 'Goals', 'A': 'Assists'})
```
